```python
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")
```

```python
import zipfile
```

```python
zip = zipfile.ZipFile("/content/brain.yolov11.zip")
```

```python
zip.extractall()
```

```python
model.train(data="/content/data.yaml" , epochs=20)
```

```python
model.save("brain.pt")
```

```python
new_model = YOLO("brain.pt")
```

```python
pred = new_model.predict("/content/test/images/1108_jpg.rf.k7rDCsrbrhkrefL8ORHx.jpg")
```

```python
pred[0].show()
```

```python
import gradio as gr
```

```python
def brain(image):
  image = new_model.predict(image)
  image = image[0].plot()
  return image
```

```python
app = gr.Interface(fn=brain,inputs=gr.Image("numpy"),outputs=gr.Image())
```

```python
app.launch()
```
